```python
import pandas as pd
df = pd.read_csv('vgsales.csv')
df.head()
```

By using df.info() i just saw all of columns and range. By non-null i just watching total number of columns and here number is same or not. If number isnot same that means have null or missing value .

```python
df.info()
```

Here i create new variable where i will put all of my new clean data. cuz before we already saww that have missingg valur so here by usng dropna i just delete all of missing value. and the  i am checking again that is there have any null if have their sum. here we can see sum for all columns is zero cuz we already clear all of missing value.

```python
df_clean = df.dropna()
df_clean.isnull().sum()
```

Now i just filter data. Here i just find top 5 games. the most selling 5 games. here I called df_clean and sorted valuesand by using ascending i just sorted it highest to lowest so in above we can see highest value. and after that by using head 5 i means i want to see top 5 sales.

```python
top_game = df_clean.sort_values(by='Global_Sales', ascending=False).head(5)
print(top_game[['Name', 'Platform', 'Global_Sales']])
```

Here i just wanted to see all infomation at 2010

```python
games_2010 = df_clean[df_clean['Year']==2010]
games_2010.head()
```

Here i wanted to see which catagory or whih types of games earn or sell the most in the history.

```python
genre_sales = df_clean.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)
print(genre_sales)
```

Here i just wanted to see which types of games get hoigest between the north america and japan. and is they were sharing same interest or not? its like reginal baesd filtering.

```python
na_genre = df_clean.groupby('Genre')['NA_Sales'].sum().sort_values(ascending=False).head(5)
print("North americar most selling games", na_genre)
```

```python
jp_genre = df_clean.groupby('Genre')['JP_Sales'].sum().sort_values(ascending= False).head(5)
print('Japan er most selling Games', jp_genre)
```

from all of platform which ine hold top place ?

```python
top_platform = df_clean.groupby("Platform")['Global_Sales'].sum().sort_values(ascending=False).head(5)
print("Most popular 5 platform", top_platform)
```

What time was golden era for selling games. here we can see...

```python
year_sale = df_clean.groupby('Year')['Global_Sales'].sum().sort_values(ascending=False).head(5)
print("The highest game selling time", year_sale)
```
